### Constructing the Domain

```python
# Define your domain here
# using following rules 0:Celeste, 1:Hollow Knight silkson, 2: It Takes Two, 3:Grounded 2, 4: None
# using the first letter of students to represent variable 
# domains_games is a dictionary 
# All students
students = ('A','B','C','D','E','F','G','H')

# Domain for each student
domains_games = { s: {0,1,2,3,4} for s in students }
print (domains_games)
```

### Constructing the Constraints: Organizing A Gaming Night

```python
# Define your constraints here
### Highlights:
# The person per game is not important because single participant can also join multiplayer games by sharing a session with other club partners.
# The constraint constructor of csp takes two arguments:
# 1. The variables that take part in the constraint
# 2. The constraint itself which is a function that takes the variables as arguments and returns true or false


# 1. Elena and Freddy both want either IT2 or Grounded 2 (same game)
# ----------------------------------------------------------------------------
def c_1(E, F):
    return E == F and E in {2, 3} #only 2 and 3 need more than 1 players
C_1 = Constraint(('E', 'F'), c_1)


# ----------------------------------------------------------------------------
# 2. No one wants to play a multiplayer game alone
# ----------------------------------------------------------------------------
def c_2(*args):
    # Count players per game
    counts = {2:0, 3:0}  # Only IT2 & G2 matter for pairing
    for a in args:
        if a in counts:
            counts[a] += 1

    return (counts[2] != 1) and (counts[3] != 1)
C_2 = Constraint(students, c_2)


# ----------------------------------------------------------------------------
# 3. Silksong (game 1) has limited spots (<4 players)
# ----------------------------------------------------------------------------
def c_3(*args):
    return sum(a == 1 for a in args) < 4
C_3 = Constraint(students, c_3)


# ----------------------------------------------------------------------------
# 4. Barney does NOT want to play solo (0,1 are false 2,3 are true if more than 2 people, 4 is true)
# ----------------------------------------------------------------------------
def c_4(A, B, C, D, E, F, G, H):
    IT2 = sum(x == 2 for x in (A,B,C,D,E,F,G,H))
    G2  = sum(x == 3 for x in (A,B,C,D,E,F,G,H))

    if B == 2:
        return IT2 >= 2
    if B == 3:
        return G2 >= 2
    if B == 0 or B == 1:
        return False
    return True   # If B chooses none 
C_4 = Constraint(students, c_4)


# ----------------------------------------------------------------------------
# 5. Elena and Gloria play together
# ----------------------------------------------------------------------------
C_5 = Constraint(('E','G'), lambda E, G: E == G)


# ----------------------------------------------------------------------------
# 6. Davin plays only IT2 or None
# ----------------------------------------------------------------------------
C_6 = Constraint(('D',), lambda d: d in {2,4})


# ----------------------------------------------------------------------------
# 7. Henry avoids Celeste
# ----------------------------------------------------------------------------
C_7 = Constraint(('H',), lambda h: h != 0)


# ----------------------------------------------------------------------------
# 8. Freddy avoids Henry’s game
# ----------------------------------------------------------------------------
C_8 = Constraint(('F','H'), lambda f, h: f != h)


# ----------------------------------------------------------------------------
# 9. Anna plays only IT2 or G2 or none 
# ----------------------------------------------------------------------------
C_9 = Constraint(('A',), lambda a: a in {2,3,4})


# ----------------------------------------------------------------------------
# 10. Barney and Claire together in Grounded 2
# ----------------------------------------------------------------------------
C_10 = Constraint(('B','C'), lambda b, c: b == 3 and c == 3)


# ----------------------------------------------------------------------------
# 11. Total budget ≤ 40
# ----------------------------------------------------------------------------
# Cost per student ONLY if they play a game (NOT None)
# Example game prices: Celeste 4, Silksong 4, IT2 6, Grounded2 5
game_cost = {0:4, 1:4, 2:6, 3:5, 4:0}

def c_11(*args):
    return sum(game_cost[a] for a in args) <= 40
C_11 = Constraint(students, c_11)


# ----------------------------------------------------------------------------
# 12. Anna, Davin, Henry play same game
# ----------------------------------------------------------------------------
C_12 = Constraint(('A','D','H'), lambda a,d,h: a == d == h)


# ----------------------------------------------------------------------------
# 13. Davin plays IT2, Elena avoids IT2
# ----------------------------------------------------------------------------
C_13 = Constraint(('D','E'), lambda d,e: d == 2 and e != 2)


# ----------------------------------------------------------------------------
# 14. No one plays alone (no game should have exactly 1 player)
# ----------------------------------------------------------------------------
def c_14(*args):
    counts = {0:0,1:0,2:0,3:0}
    for a in args:
        if a in counts:
            counts[a] += 1
    return all(v != 1 for v in counts.values())
C_14 = Constraint(students, c_14)


# ----------------------------------------------------------------------------
# 15. Grounded 2 must have ≥3 players
# ----------------------------------------------------------------------------
C_15 = Constraint(students, lambda *args: sum(a == 3 for a in args) >= 3)


# ----------------------------------------------------------------------------
# 16. At least 3 different games are chosen
# ----------------------------------------------------------------------------
def c_16(*args):
    chosen = {a for a in args if a != 4}
    return len(chosen) >= 3
C_16 = Constraint(students, c_16)
```
