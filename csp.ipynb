### Constructing the Constraints: Organizing A Gaming Night

```python
# Define your constraints here
### Highlights:
# The person per game is not important because single participant can also join multiplayer games by sharing a session with other club partners.
# The constraint constructor of csp takes two arguments:
# 1. The variables that take part in the constraint
# 2. The constraint itself which is a function that takes the variables as arguments and returns true or false
# 1. Elena and Freddy want to play together
def c_1(*args):
    return args[0] == args[1] and args[0] in {2, 3}
C_1 = Constraint(('E','F'), c_1)

# 2. No one wants to share a game session with strangers
def c_2(*args):
    IT2 = sum(1 for arg in args if arg==2)
    G2 = sum(1 for arg in args if arg==3)
    return IT2 % 2 == 0 and G2 % 2 == 0
C_2 = Constraint(('A','B','C','D','E','F','G','H'), c_2)

# 3. Hollow Knight: Silksong limited spots
def c_3(*args):
    Hol = sum(1 for arg in args if arg==1)
    return Hol < 4
C_3 = Constraint(('A','B','C','D','E','F','G','H'), c_3)

# 4. Barney does not want to play solo
def c_4(*args):
    IT2 = sum(1 for arg in args if arg==2)
    G2 = sum(1 for arg in args if arg==3)
    B = args[1]  # B's choice
    return ((B==2 and IT2>1) or (B==3 and G2>1))
C_4 = Constraint(('A','B','C','D','E','F','G','H'), c_4)

# 5. Elena and Gloria together
def c_5(*args):
    return args[0]==args[1]
C_5 = Constraint(('E','G'), c_5)

# 6. Davin plays only It Takes Two-
C_6 = Constraint(('D',), lambda d: d==2)

# 7. Henry avoids Celeste
C_7 = Constraint(('H',), lambda h: h != 0)

# 8. Freddy avoids Henry
C_8 = Constraint(('F','H'), lambda f,h: f != h)

# 9. Anna plays only It Takes Two or Grounded 2
C_9 = Constraint(('A',), lambda a: a in {2,3})

# 10. Barney and Claire together in Grounded 2
C_10 = Constraint(('B','C'), lambda b,c: b==3 and b==c)

# 11. Total budget <= 40
def c_11(*args):
    costs = [4,4,6,5,0,0,0,0]  # example indices for students
    Budget = sum(cost for cost,arg in zip(costs,args) if arg !=0)
    return Budget <= 40
C_11 = Constraint(('A','B','C','D','E','F','G','H'), c_11)

# 12. Anna, Davin, Henry together
C_12 = Constraint(('A','D','H'), lambda a,d,h: a==d==h)

# 13. Davin It Takes Two, Elena avoids puzzle
C_13 = Constraint(('D','E'), lambda d,e: d==2 and e!=d)

# 14. No one plays alone
def c_14(*args):
    counts = {0:0, 1:0, 2:0, 3:0}
    for arg in args:
        if arg in counts: counts[arg] += 1
    return all(v!=1 for v in counts.values())
C_14 = Constraint(('A','B','C','D','E','F','G','H'), c_14)

# 15. Grounded 2 requires >=3 players
def c_15(*args):
    G2 = sum(1 for arg in args if arg==3)
    return G2 >=3
C_15 = Constraint(('A','B','C','D','E','F','G','H'), c_15)

# 16. At least 3 different games
def c_16(*args):
    games = {0,1,2,3}
    chosen = {g for g in games if g in args}
    return len(chosen) >= 3
C_16 = Constraint(('A','B','C','D','E','F','G','H'), c_16)


```
